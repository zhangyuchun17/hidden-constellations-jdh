Queer sapphic spatial practices in New York City leave uneven documentary traces that do not readily resolve into point-based geospatial datasets. This article asks a historical question about that unevenness: under what documentary conditions does sapphic place-knowledge become addressable enough to sustain repeatable, map-ready locational claims. To answer this, the project compares three New York City corpora produced under distinct documentary regimes and constructs an address-centered corpus with a minimal shared schema while keeping provenance visible rather than consolidating records into a single authoritative geography. It contributes a transferable form of spatial source criticism by making addressability auditable and by retaining non-pointability and geocoding outcomes as interpretable evidence of changing legibility conditions.

## Introduction: The Evidentiary Persuasion of Point Maps

Point maps have become one of the most persuasive genres in digital spatial history. A dot appears to fix a place, imply repeatability, and invite comparison across time. Over the last two decades, LGBTQ spatial histories have increasingly been written through such maps, drawing on digital heritage inventories, community directories, commercial listings, and scholarly reconstructions to assemble address-based corpora of venues across cities and decades. Bar histories in particular have been reconstructed through archival research and business records, yielding influential accounts of nightlife ecologies, police regulation, and community formation (<cite id="l1n2e"><a href="#zotero%7C23211570%2F8Z7TNWBQ">(Faderman, 1981)</a></cite>, <cite id="a8pt3"><a href="#zotero%7C23211570%2FJSTJ8HPV">(Mattson, 2019)</a></cite>, <cite id="drl0n"><a href="#zotero%7C23211570%2FETBUK2QK">(Stegall, Gwendolyn, 2019)</a></cite>). These projects have expanded what can be located, counted, and visualized in the historical record.

Yet the map’s persuasive clarity can also mislead. When dots thicken in one decade and thin in another, the shift invites an immediate reading as rise and decline, emergence and disappearance. Such readings often depend on evidence that already meets a spatial precondition: it must be nameable in a publishable way, stabilizable into an address-like locator, and translatable into a machine-actionable input that a geocoder can resolve. The apparent stability of the point is therefore not only a representational choice; it is also an evidentiary achievement produced by documentary infrastructures. What appears as a neutral “location” is frequently the endpoint of prior selection, standardization, and loss.

Critical historical GIS has repeatedly warned against allowing coordinate precision to stand in for evidentiary strength. Work in the spatial humanities has emphasized that historical spatial evidence is frequently expressive, contingent, and uneven, and that the task is not simply to convert such evidence into points but to retain and make visible its limits and forms (<cite id="kd2b8"><a href="#zotero%7C23211570%2FMBPU49QA">(Knowles &#38; Hillier, 2008)</a></cite>, <cite id="a2exn"><a href="#zotero%7C23211570%2FZ5VAUWTY">(Bodenhamer et al., 2010)</a></cite>). From this perspective, point maps do not simply show where the past happened; they show what kinds of past traces became addressable enough to be indexed as locational claims.

This article approaches queer sapphic spatial histories as a case where the costs of that evidentiary compression become especially visible. Lesbian historiography has long shown that women’s same-sex attachments are unevenly named and variably recorded across the nineteenth and twentieth centuries, often surfacing through indirect idioms rather than stable public self-designation (<cite id="t3zoo"><a href="#zotero%7C23211570%2FNHNG4MT8">(Vicinus, 1992)</a></cite>). As a result, sapphic place-knowledge often circulates through heterogeneous locators (e.g., intersections, corridors, neighborhood references, program-specific labels, or name-only mentions) that are meaningful within communities yet resistant to address-level stabilization. What looks like sparse mapping may therefore reflect not only social change but also shifts in what could be safely named, widely published, and repeatedly indexed.

Read this way, the point map becomes a selective archive. Certain traces arrive as full addresses; others persist as relational descriptions. The difference is not a matter of data quality alone. It indicates the boundary conditions under which place can become repeatably addressable as point geometry. The historical problem that follows is therefore not primarily how to complete an inventory, but how to read mapped density and mapped thinning as the joint outcome of social formation and documentary addressability.

### Evidentiary thresholds, addressability, and accountable mapping

The guiding question is prior to any inventory: under what documentary conditions does queer place-knowledge become addressable enough to sustain repeatable, map-ready locational claims? To examine this, the article compares three New York City corpora produced under distinct documentary regimes: a heritage inventory (<cite id="25hpc"><a href="#zotero%7C23211570%2FGF49PUCC">(<i>NYC LGBTQ Historic Sites Project</i>, n.d.)</a></cite>), a community-driven directory (<cite id="lrqun"><a href="#zotero%7C23211570%2FBBLHTWMR">(<i>ADDRESSES PROJECT</i>, n.d.)</a></cite>), and a scholarly reconstruction of lesbian bar history (<cite id="2u83u"><a href="#zotero%7C23211570%2FETBUK2QK">(Stegall, Gwendolyn, 2019)</a></cite>). Rather than consolidating heterogeneous records into a single authoritative geography, the project constructs an address-centered corpus with a minimal shared schema while keeping provenance visible.

Spatial claims are handled by locational form, with an audit layer that preserves provenance and retains non-pointability and geocoding outcomes as evidence rather than coercing all traces into point geometry. This design follows the reading rules by keeping non-pointability and geocoding outcomes inside the evidentiary surface rather than repairing them away.

By foregrounding evidentiary gating, documentary regimes, and spatial precision classes, the article reframes decade-based clusters as documentary events and point-level thinning as prompts for spatial source criticism. The result is a comparative evidentiary surface for debating changing legibility conditions across regimes, rather than a definitive geography of sapphic space.

## Definitions

### Sapphic

In this article, sapphic operates as a methodologically motivated umbrella rather than a retrospective identity assignment. The term names a historiographic problem before it names a population: across the nineteenth and twentieth centuries, the social formations this project seeks to locate were rarely self-designated in stable, publicly circulatable terms. Relevant attachments, affiliations, and community formations became legible through indirect or contextually specific idioms whose naming conventions shifted across decades and documentary contexts (<cite id="t3zoo"><a href="#zotero%7C23211570%2FNHNG4MT8">(Vicinus, 1992)</a></cite>). 

This deliberate looseness is circumscribed: it concerns the framing of actor identity in sources, not a claim that all entries share equivalent social meaning. It is meant to keep open what a given source was actually recording about social affiliation, rather than forcing past formations to pass through a single contemporary label as an entry condition. Methodologically, the umbrella reduces the risk that the corpus will reproduce the classificatory limits of listing infrastructures by only capturing actors who were already named in institutionally searchable ways. It also keeps “label drift” interpretable as historical evidence: shifts in naming are treated as part of what conditions indexability and, by extension, what can be mapped.

### Queer sapphic space

Queer sapphic space is defined here as historically situated place-knowledge through which sapphic affiliation and recognition become meaningful within particular social and documentary contexts (<cite id="aq7fq"><a href="#zotero%7C23211570%2FH8A3MZPK">(Massey, 2009)</a></cite>, <cite id="3dcmc"><a href="#zotero%7C23211570%2FAQCJQA9Y">(Berlant &#38; Warner, 1998)</a></cite>). 

The analysis focuses on when fragments of this place-knowledge become recordable as locatable claims under specific documentary regimes.

### Addressability and recordability

Addressability refers here to the historically contingent process by which sapphic place-knowledge becomes recordable as a locatable claim. Addressability includes not only street addresses but also other locational forms that circulate in sources (intersections, corridors, neighborhood references, or name-only mentions). Coordinates are treated as one possible output of addressability, not its definition. Records that cannot support point precision, as well as geocoding failures, are retained as classified outcomes.

In this article, geocoding is treated as a conditional test of address-level claims rather than as a default step of completion.

### Documentary regimes

Documentary regimes refer to patterned conditions of evidence-making: the purposes of records, naming categories, verification norms, survivable publicity, and institutional preservation pathways that shape what becomes locatable. 

The article compares regimes to interpret mapped clustering and thinning through both recordability conditions and social concentration, treating neither as sufficient on its own.

## Documentary Regimes and Evidentiary Rules

### Corpora and regimes

The address-centered corpus is assembled from three New York City–focused corpora produced under distinct documentary regimes. 

The NYC LGBT Historic Sites Project is used as a heritage-oriented corpus of publicly documented sites. Entries typically provide narratable site descriptions and, in many cases, address-level locators suitable for repeatable citation. The Addresses Project is used as a community directory corpus compiled through submissions and community-transmitted place knowledge. Entries are organized in an address-centered format and often include locational cues that can be aligned to the shared schema. Stegall’s reconstruction of lesbian bar history (<cite id="2u83u"><a href="#zotero%7C23211570%2FETBUK2QK">(Stegall, Gwendolyn, 2019)</a></cite>) is used as a scholarly synthesis corpus that contributes historically contextualized nightlife references across decades. Because coordinates are not consistently provided, this corpus frequently requires present-day spatialization when address-level locators are available. 

Municipal spatial layers, including basemaps and neighborhood boundaries from NYC Open Data, are used exclusively as reference layers for visualization and spatial validation and do not constitute evidentiary claims.

Together, these corpora establish the project’s comparative frame for assessing regime-specific recordability. They represent three distinct documentary formats through which place-knowledge becomes recordable as mappable locational claims.

### Reading rule

Three interpretive rules guide the reading of the compiled corpus. First, harmonization produces schema comparability without collapsing heterogeneous records into a single authoritative site; overlap across sources is treated as evidentiary reinforcement rather than as grounds for consolidation. Second, mapped clustering and thinning are interpreted in relation to the documentary regimes that make such patterns operable in a given decade, keeping provenance visible in all comparisons. Third, non-pointable references and geocoding failures are retained as outcome categories rather than repaired away, marking the boundary between historical trace and machine-readable address. These rules ensure that the corpus remains a comparative evidentiary surface rather than a consolidated truth map.

## Method: A Layered Evidentiary Workflow

The workflow proceeds in four stages. First, records from the three NYC corpora are ingested without consolidation and aligned to a minimal shared schema that retains source provenance and core descriptive fields, producing a harmonized core table for comparison. Second, temporal signals are normalized to a decade scale using structured start/end years, bounded date ranges, or reported decades, with the temporal basis recorded rather than inferred when missing. Third, spatialization is performed conditionally: entries that already contain coordinates are retained as given, while entries lacking coordinates are routed by locational form. Only records that provide a street address enter automated geocoding; references expressed as intersections, corridors, neighborhood-level descriptions, or name-only mentions are retained as non-pointable outcomes and assigned spatial precision classes. Fourth, an evidence log records spatial precision, geocoding status (matched, skipped, no-match), and manual review traces, linking each decision to its provenance and rationale. The outputs include a harmonized core corpus, a geocoded subset where address-level claims support point geometry, and an audit layer that makes non-pointability and geocoding failures inspectable alongside decade-based visualizations.

### Schema harmonization and temporal normalization

Records from different sources are harmonized through a small set of shared fields that preserve provenance while enabling cross-source comparison. These include canonicalized place names, site-type tags, source references, temporal signals, and spatial-precision indicators. Harmonization does not resolve discrepancies between sources; instead, it standardizes how those discrepancies are recorded and made comparable.

Temporal information is normalized to a decade scale based on structured fields such as reported decades, start and end years, or bounded date ranges. This normalization supports a timeline-based reading that foregrounds documentary rhythm and density rather than smoothing temporal uncertainty into falsely precise year-level claims. Where no structured temporal signal is available, time precision is explicitly marked as missing rather than inferred from narrative description.

```python
def finalize_decades_and_basis(
    reported_decades: pd.Series,
    derived_decades: pd.Series,
    source_tag: str
) -> tuple[pd.Series, pd.Series]:

    decades_norm = np.where(
        reported_decades.map(len) > 0,
        reported_decades.map(json.dumps),
        np.where(
            derived_decades.map(len) > 0,
            derived_decades.map(json.dumps),
            json.dumps([])
        )
    )

    time_basis = np.select(
        [
            reported_decades.map(len) > 0,
            (reported_decades.map(len) == 0) & (derived_decades.map(len) > 0)
        ],
        [
            f"{source_tag}.decades_reported",
            f"{source_tag}.decades_derived_from_bounds"
        ],
        default=f"{source_tag}.decades_missing"
    )

    return decades_norm, time_basis
```

### Spatialization, geocoding, and limits of point precision

Spatialization follows explicit decision rules that distinguish between addressable and non-addressable location claims.

Only records that provide a verifiable street address are eligible for automated geocoding. If coordinates are already supplied by the source, they are retained with provenance. If a street address is present but no coordinates are provided, the record enters the geocoding pipeline and receives a status flag (matched, no_match, or error). Automated results are stored as returned and are not silently overwritten.

Records expressed through intersections, corridor descriptions, neighborhood-level references, or name-only mentions do not enter automated geocoding. These entries are retained as valid spatial claims and assigned spatial-precision classes without being coerced into point geometry. Their non-pointability is treated as an outcome category rather than as missing data.
Manual spatial verification is applied selectively. When a geocoded street address yields ambiguous or historically inconsistent results (e.g., known street renaming, venue relocation, or multi-source conflict), a manual check may confirm, revise, or withhold coordinates. Any manual intervention is logged with the original value, the revised value if applicable, the rationale for the decision, and the source consulted.

These rules ensure that spatialization operates as a bounded interpretive procedure. Address-level precision is produced only where documentary form supports it, while non-addressable references and geocoding failures remain inspectable as evidence of recordability limits.

Raw location form | Precision class | Geocoded? | Geometry generated | Evidentiary status
---|---|---|---|---
Full street address | point | Yes | Yes (point) | Addressable
Street intersection | intersection | No | No | Valid non-point claim
Between X and Y | street_segment | No | No | Corridor reference
Neighborhood reference | neighborhood | No | Optional area | Approximate claim
Name-only reference | descriptive | No | No | Memory trace
Missing / unverifiable | unresolved | Attempted | No | Machine-limit evidence

```python
import re
from typing import Optional


# 1. Classify address type (not all locations are pointable)
def classify_location(raw_address: Optional[str]) -> str:
    """
    Distinguish between addressable and non-pointable spatial claims.
    """
    if raw_address is None:
        return "missing"

    s = raw_address.lower()

    if "between" in s:
        return "street_segment"
    if " and " in s or " & " in s:
        return "intersection"
    if re.match(r"^\d", s):
        return "street_address"

    return "descriptive"


# 2. Decide whether geocoding is appropriate
def should_geocode(location_type: str) -> bool:
    """
    Only stable street addresses are sent to geocoding.
    Other spatial claims remain valid but non-pointable.
    """
    return location_type == "street_address"


# 3. Record geocoding outcome without discarding failure
def record_geocode_result(match_status: str) -> str:
    """
    Geocoding failure is preserved as evidence,
    not silently corrected.
    """
    if match_status == "Match":
        return "point"
    else:
        return "unresolved"


# 4. Handle multi-source convergence
def record_convergence(source_count: int) -> str:
    """
    Convergence strengthens evidentiary status
    without merging records at construction stage.
    """
    if source_count > 1:
        return "multi_source_candidate"
    return "single_source"
```

### Making uncertainty auditable: evidence log as a method of spatial source criticism

The evidence log records how each record’s temporal and spatial claims were stabilized, bounded, or left unresolved. It keeps three kinds of information visible. First, it preserves provenance by retaining source identifiers and distinguishing between source wording and canonicalized forms. Second, it records the basis and limits of spatialization. Entries are classified by what kind of locational claim they make, and by what level of precision they can responsibly sustain.  Third, it stores review decisions. When records conflict across sources, when matches are ambiguous, or when a site appears as a multi-source candidate, manual audit is triggered and the rationale for any intervention is documented.

This protocol treats uncertainty as part of the evidentiary surface. Some records remain non-pointable, some remain temporally coarse, and some are kept as unresolved traces. Rather than forcing these cases into point geometry, the workflow preserves them as classified outcomes that mark the boundary between historical trace and machine-readable address. The evidence log thus functions as a form of spatial source criticism: it makes the conditions under which mappable claims are produced explicit, and it keeps mapping accountable to its own spatial precision classes.

```python
import json
import pandas as pd
import numpy as np


def _json_list(x):
    """Parse decades_norm stored as JSON list string; return [] if missing/unparseable."""
    if x is None or (isinstance(x, float) and np.isnan(x)) or str(x).strip() == "":
        return []
    try:
        v = json.loads(str(x))
        return v if isinstance(v, list) else []
    except Exception:
        return []


def derive_time_precision_class(df: pd.DataFrame) -> pd.Series:
    """
    decade      : decades_norm is a non-empty list
    year_range  : start_year or end_year exists but decades_norm empty
    missing     : no structured time signals
    """
    decades = df["decades_norm"].apply(_json_list)
    has_decade = decades.apply(lambda v: len(v) > 0)

    start = pd.to_numeric(df["start_year"], errors="coerce")
    end   = pd.to_numeric(df["end_year"], errors="coerce")
    has_year_bounds = start.notna() | end.notna()

    out = pd.Series(["missing"] * len(df), index=df.index)
    out.loc[has_decade] = "decade"
    out.loc[(~has_decade) & has_year_bounds] = "year_range"
    return out


def derive_spatial_precision_class(df: pd.DataFrame) -> pd.Series:
    """
    point          : lat/lon present
    intersection   : address_kind == 'intersection'
    non_pointable  : between_phrase / descriptive / multi_address
    missing        : none of the above
    """
    lat = pd.to_numeric(df["lat"], errors="coerce")
    lon = pd.to_numeric(df["lon"], errors="coerce")
    has_coords = lat.notna() & lon.notna()

    kind = df["address_kind"].fillna("missing").astype(str).str.lower()

    out = pd.Series(["missing"] * len(df), index=df.index)
    out.loc[kind.eq("intersection")] = "intersection"
    out.loc[kind.isin(["between_phrase", "descriptive", "multi_address"])] = "non_pointable"
    out.loc[(out.eq("missing")) & has_coords] = "point"
    return out


def validate_audit_trace(df: pd.DataFrame) -> pd.DataFrame:
    """
    Minimal JDH-style validation:
    - prevent over-pointification
    - require complete geocode trace when attempted
    """
    issues = []

    kind = df["address_kind"].fillna("missing").astype(str).str.lower()

    # (1) geocode_attempted implies geocode_status recorded (trace completeness)
    attempted = df["geocode_attempted"].fillna(False).astype(bool)
    status_missing = df["geocode_status"].isna() | (df["geocode_status"].astype(str).str.strip() == "")
    bad = attempted & status_missing
    for i in df.index[bad]:
        issues.append({"severity": "error", "entry_id": df.at[i, "entry_id"],
                       "issue": "geocode_attempted_without_status"})

    # (2) spatial_precision_class=point but address_kind non-pointable (over-pointification)
    non_pointable = kind.isin(["between_phrase", "descriptive", "multi_address"])
    bad = df["spatial_precision_class"].eq("point") & non_pointable
    for i in df.index[bad]:
        issues.append({"severity": "error", "entry_id": df.at[i, "entry_id"],
                       "issue": "over_pointification"})

    return pd.DataFrame(issues)
```

### Non convergence as evidence: classification instability and multi phase lineages

A second finding concerns the instability of classification when the same vicinity enters the corpus through different documentary regimes. In an address based workflow, it is tempting to treat such divergence as an error to be resolved by choosing a single label, a single venue identity, or a single authoritative description. Read as spatial source criticism, however, non convergence is itself an evidentiary condition. It shows that a place can become sapphic in historically bounded ways without stabilizing as a continuously nameable object in address based records, and it clarifies why regime visibility must remain preserved rather than repaired away.

One midtown nightlife venue provides a concrete illustration. In Stegall’s reconstruction, the venue became legible through women-only programming on specific nights in the early 1990s. Its sapphic quality is therefore produced under a programmatic logic that treats temporal restriction and entry rules as constitutive of what the place is for, and for whom. In a record from the Addresses Project, the same vicinity is framed differently, appearing as a drag oriented venue on Eleventh Avenue in the mid 40s. Here the place becomes recordable through another public facing cultural idiom that is historically adjacent yet not equivalent, and the addressable  unit is shaped by what the directory can name and circulate. Later public accounts of the same site are structured primarily through enforcement narratives, where the venue is rendered legible as a target of regulation or policing rather than as a community described space. Taken together, these traces suggest that a space may become sapphic conditionally, by weekday, by event, or by entry rule, while never stabilizing as a single continuously listed lesbian bar. What appears as a discrepancy in a consolidated dataset is better read as a multi phase lineage whose meanings are produced through different documentary thresholds.

For this reason, the workflow does not force cross source convergence into a single venue category. Convergence would require choosing which regime defines the identity of the place, and such a choice would collapse the very historical variability that the corpus is meant to render comparable. Instead, provenance fields, corroboration markers, and interpretive notes preserve parallel registrations so that classification instability remains visible at the level of evidence. This makes the divergence reviewable rather than accidental. Readers can trace how the same vicinity is stabilized differently across sources, which temporal bases are implied, and which naming conventions and publication logics are at work. In this sense, the evidence log is not ancillary documentation. It is the interpretive surface that allows the project to treat instability as a result that can be checked and debated, rather than as noise to be cleaned away. 

This finding also clarifies why recordability can thin in patterned ways that cannot be attributed to social absence. Mattson’s account of national guides and listings provides one mechanism. For much of the period in which commercial bar guides circulated, their production and categorization practices were shaped by teams and infrastructures that did not prioritize women centered sociality, and women focused guides emerged comparatively late. The implication for an address centered timeline is not simply that fewer women oriented places existed, but that fewer were routed into market facing listing infrastructures that could stabilize them as repeatable address level claims. In that sense, documentary thinning can reflect a weak zone in the circulation and indexing of women centered nightlife, even where programmatic, episodic, or selectively disclosed formations persisted. Read together with the midtown case, the broader point is that regime specific naming and publication pathways shape what becomes listable, and listability in turn shapes what becomes mappable. Non convergence therefore matters as evidence. It marks where sapphic place knowledge is produced through contingent framings and multi phase lineages that an address based category system cannot responsibly collapse into a single continuously countable object.

### Precision structure and geocoding outcomes as historical signals

The address centered corpus does not only vary in how many entries can be mapped in a given decade. It also varies in what kinds of location claims the evidence can sustain. A substantial portion of traces enters the workflow as intersections, street segments expressed through between descriptions, corridor scale references, neighborhood level mentions, or name only sites. These forms are treated here as legitimate spatial claims rather than as errors, because they are the forms in which place knowledge was recorded, circulated, and made publishable under particular documentary regimes. Their persistence across decades indicates that addressability is not equivalent to point geometry. Point precision is one possible output of addressability, but it is not its definition, and it cannot be assumed as the default without erasing the evidentiary form of the trace. For this reason, the workflow bounds spatialization through explicit precision classes and records them in the evidence log, so that readers can see when a record supports an address level point, when it supports only a corridor scale claim, and when it remains a nameable but non pointable reference. Figure 7 makes this structure visible by showing how mapped precision shifts by decade, and by demonstrating that the evidentiary surface includes stable non point outcomes alongside point locations.

Within this framework, unmapped records are not treated as outside the analysis. They are treated as an outcome category produced by the same decision rules that govern point stabilization. Figure 8 separates unmapped entries by decade and by source, allowing the reader to compare where non pointability concentrates across regimes and periods. This view supports two kinds of interpretation that are not available in a point only map. First, it reveals that some documentary regimes more often yield location claims that do not resolve into street address points, either because they were recorded in descriptive forms, or because the regime privileged publishable narratives over repeatable address listings. Second, it shows that the reliance on corridor scale and neighborhood scale references is historically uneven. In some decades and districts, the trace is more likely to circulate through vague or relational locators rather than through full addresses, which means that a thinning of point maps can reflect a shift in record format and indexability rather than a simple decline in social practice. Under this reading rule, unmapped is not a synonym for absence. It is a signal of where legibility reaches the boundary of what can be responsibly stabilized as an address level claim, and it therefore functions as a diagnostic surface for spatial source criticism.
These boundaries are historical rather than incidental. They mark moments when documentary regimes and cartographic conventions encounter the limits of responsible stabilization, and they show how the archive and the geocoder jointly shape what becomes countable and comparable.

## Discussion and Conclusion

### Discussion: What decade maps can and cannot be read to show

The findings above clarify how decade-based point maps should be read in this case. Across heterogeneous documentary regimes, mapped density is not a transparent proxy for social concentration. It is best treated as a composite signal produced where social gathering becomes documentarily stabilizable as an address-level claim. Under this reading rule, a cluster can register repeated use of a place, but it can also register the arrival of a record type, an inventory practice, or a category vocabulary that makes that kind of place newly countable. Conversely, thinning and blankness do not straightforwardly imply diminished sapphic presence. They often indicate that place knowledge circulated in locational forms that do not reliably survive as repeatable address-based claims.

This shift in interpretive stance also changes the purpose of integration. The aim of assembling multiple corpora is not to correct partial archives into a single baseline geography. It is to maintain a comparative evidentiary frame in which sources remain distinguishable, and in which overlap and divergence can be read as properties of evidence production. In that frame, disagreement is not a technical defect to be resolved by consolidation. It is a prompt for historical inquiry into how different documentary regimes stabilize different kinds of places, with different verification norms and naming conventions, and therefore produce different map-ready surfaces.

### Limits and scope conditions: What an address-based evidentiary frame cannot capture

This article remains bounded by the documentary regimes it can ethically and practically assemble. The compiled corpus inherits survivorship bias, uneven temporal coverage, and uneven spatial reach across districts and sub-scenes. These constraints are not incidental. They are part of the historical conditions through which address-based evidence becomes available, and they therefore shape what can be argued from the mapped surface.

Interpretive judgment is also unavoidable when temporal and locational references are partial, historically unstable, or intentionally vague. The evidence log does not remove that judgment. It makes it inspectable by linking each plotted or unplotted record to its provenance, its temporal basis, and its stated precision limit. Auditability here is not a claim to objectivity. It is a commitment to making the production of locational claims explicit enough to be reviewed and contested.

Finally, an address-based frame necessarily under-represents relational and non-disclosable spatial practices. Many forms of queer sapphic presence are enacted through discretion, episodic coordination, and context-dependent disclosure, and these formations may not stabilize into repeatable address-like entries. This is not a gap to be solved by more aggressive geocoding. It is a scope condition that marks where complementary qualitative and participatory approaches become necessary when the question shifts from recordability to lived spatial meaning.

### Conclusion: Two contributions and a transferable template for spatial source criticism

First, this article contributes a historiographic mechanism for interpreting point-based spatial history under uneven documentation. It shows that addressability is a historically contingent evidentiary condition and that the map’s apparent stability is an achievement produced by documentary infrastructures, not a neutral baseline.

Second, the article contributes a transferable reporting template for spatial source criticism that makes mapped evidence answerable to its thresholds. Rather than offering a pipeline to replicate, it translates the reading rules into a reusable practice of reporting. The template has three moves: preserve regime visibility by keeping provenance explicit and treating overlap as reinforcement rather than as grounds for consolidation; bound spatial precision by classifying what kind of locational claim each record makes and stating the highest point precision it can responsibly sustain; and keep non-pointable references and geocoding failures inside the evidentiary surface by recording them as outcome categories under explicit decision rules. An evidence log implements this template by linking each locational outcome to its documentary basis, temporal basis, precision class, and review trace. The result is a map-ready corpus whose limits remain legible and whose cross-source comparisons do not require collapsing heterogeneous traces into a single authoritative geography.
